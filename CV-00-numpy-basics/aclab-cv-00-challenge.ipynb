```python
import os
import numpy as np
import matplotlib.pyplot as plt
```

## Student code

Replace `image = None` with your implementation.
Keep the variable name `image`.


```python
HEIGHT = 400
WIDTH = 400

# ==============================
# STUDENT CODE
# ==============================

# Step 1: Create a 400 x 400 RGB image.
BG_COLOR = 240
RED = [220, 60, 60]
GREEN = [60, 160, 90]
BLUE = [50, 90, 200]
YELLOW = [225, 170, 55]
# Step 2: Add at least three visible shapes or regions
# using NumPy indexing or slicing.

# YOUR CODE HERE
# nền xám nhạt, shape (400, 400, 3), dtype uint8
image = np.full((HEIGHT, WIDTH, 3), 240, dtype=np.uint8)

# 4 vùng màu bằng slicing image[y1:y2, x1:x2]

image[50:150, 50:150] = RED # hình vuông đỏ
image[180:240, 40:320] = GREEN # thanh xanh lá ngang
image[60:240, 330:370 ] = BLUE # thanh xanh dương đứng
image[280:330, 60:220] = YELLOW # thanh màu vàng ngang
```

## Validation


```python
if image is None:
    print(
        "Challenge not completed yet. "
        "Complete the STUDENT CODE section first."
    )
else:
    assert isinstance(image, np.ndarray),         "image must be a NumPy array."

    assert image.shape == (400, 400, 3),         f"Expected shape (400, 400, 3), got {image.shape}."

    assert image.dtype == np.uint8,         f"Expected dtype uint8, got {image.dtype}."

    unique_colors = np.unique(image.reshape(-1, 3), axis=0)

    assert len(unique_colors) >= 4,         "Use a background and at least three visible colors/regions."

    print("Basic checks passed.")
```

## Final output


```python
if image is not None:
    plt.figure(figsize=(6, 6))
    plt.imshow(image)
    plt.title("CV-00 Final Output")
    plt.axis("off")
    plt.show()
```
